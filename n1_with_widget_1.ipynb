شغّل الخلية دي بعد تشغيل كل خلايا التدريب اللي فوق (لازم `model`, `x`, `x_test`, `y_test` يكونوا موجودين في الذاكرة).

```python
# لو ipywidgets مش متثبتة: pip install ipywidgets

import ipywidgets as widgets
from IPython.display import display, clear_output, HTML
import numpy as np
import pandas as pd

feature_names = list(x.columns)
N_FEATURES = len(feature_names)

# ---------- CSS احترافي ----------
display(HTML(f"""
<style>
.sonar-card {{
    background: linear-gradient(135deg, #0f172a 0%, #1e293b 100%);
    border-radius: 16px;
    padding: 24px 28px;
    color: #e2e8f0;
    font-family: 'Segoe UI', Tahoma, sans-serif;
    box-shadow: 0 8px 24px rgba(0,0,0,0.35);
    margin-bottom: 12px;
}}
.sonar-title {{
    font-size: 22px;
    font-weight: 700;
    background: linear-gradient(90deg, #38bdf8, #818cf8);
    -webkit-background-clip: text;
    -webkit-text-fill-color: transparent;
    margin-bottom: 4px;
}}
.sonar-subtitle {{
    font-size: 13px;
    color: #94a3b8;
    margin-bottom: 0px;
}}
.sonar-badge {{
    display: inline-block;
    padding: 3px 10px;
    border-radius: 999px;
    font-size: 11px;
    font-weight: 600;
    background: rgba(56,189,248,0.15);
    color: #38bdf8;
    border: 1px solid rgba(56,189,248,0.4);
    margin-top: 6px;
    margin-right: 6px;
}}
.result-box {{
    border-radius: 14px;
    padding: 18px 22px;
    margin-top: 10px;
    font-family: 'Segoe UI', Tahoma, sans-serif;
}}
.result-rock {{
    background: rgba(56,189,248,0.10);
    border: 1px solid rgba(56,189,248,0.35);
}}
.result-mine {{
    background: rgba(248,113,113,0.10);
    border: 1px solid rgba(248,113,113,0.4);
}}
.result-label {{
    font-size: 20px;
    font-weight: 700;
    margin-bottom: 10px;
    color: #f1f5f9;
}}
.prob-row {{
    display: flex;
    align-items: center;
    margin-bottom: 8px;
    font-size: 13px;
    color: #cbd5e1;
}}
.prob-name {{ width: 110px; }}
.prob-bar-bg {{
    flex: 1;
    background: rgba(148,163,184,0.15);
    border-radius: 6px;
    height: 10px;
    overflow: hidden;
    margin: 0 10px;
}}
.prob-bar-fill-r {{ height: 100%; background: linear-gradient(90deg,#38bdf8,#0ea5e9); }}
.prob-bar-fill-m {{ height: 100%; background: linear-gradient(90deg,#f87171,#ef4444); }}
.prob-val {{ width: 55px; text-align: right; font-weight: 600; color: #f1f5f9; }}
.warn-box {{
    background: rgba(250,204,21,0.12);
    border: 1px solid rgba(250,204,21,0.4);
    border-radius: 10px;
    padding: 12px 16px;
    color: #fde68a;
    font-family: 'Segoe UI', Tahoma, sans-serif;
    font-size: 13px;
}}
.info-box {{
    background: rgba(56,189,248,0.10);
    border: 1px solid rgba(56,189,248,0.3);
    border-radius: 10px;
    padding: 12px 16px;
    color: #bae6fd;
    font-family: 'Segoe UI', Tahoma, sans-serif;
    font-size: 13px;
}}
</style>

<div class="sonar-card">
    <div class="sonar-title">**7r7a4\n🔍 Sonar Signal Classifier ** </div>
    <div class="sonar-subtitle">تصنيف إشارات السونار: صخرة (Rock) أو لغم (Mine) — باستخدام Logistic Regression</div>
    <div style="margin-top:10px;">
        <span class="sonar-badge">{N_FEATURES} Features</span>
        <span class="sonar-badge">Trained Model</span>
    </div>
</div>
"""))

# ---------- عناصر التحكم ----------
textbox = widgets.Textarea(
    value='',
    placeholder=f'الصق هنا {N_FEATURES} قيمة مفصولة بفواصل تمثل قراءات السونار ...',
    layout=widgets.Layout(width='100%', height='90px', border='1px solid #334155'),
)

random_btn = widgets.Button(
    description="🎲  عينة عشوائية",
    button_style='info',
    layout=widgets.Layout(width='170px', height='38px'),
)
predict_btn = widgets.Button(
    description="🚀  تنبؤ الآن",
    button_style='success',
    layout=widgets.Layout(width='170px', height='38px'),
)
clear_btn = widgets.Button(
    description="🗑️  مسح",
    button_style='warning',
    layout=widgets.Layout(width='120px', height='38px'),
)

buttons_row = widgets.HBox(
    [random_btn, predict_btn, clear_btn],
    layout=widgets.Layout(justify_content='flex-start', margin='10px 0 0 0'),
)

output = widgets.Output()

controls_box = widgets.VBox(
    [textbox, buttons_row],
    layout=widgets.Layout(
        border='1px solid #334155',
        padding='16px',
    ),
)

display(controls_box, output)


# ---------- المنطق ----------
def load_random_sample(_):
    idx = np.random.randint(0, len(x_test))
    sample = x_test.iloc[idx]
    textbox.value = ', '.join(map(str, sample.values))
    with output:
        clear_output()
        display(HTML(
            f'<div class="info-box">✅ تم تحميل عينة رقم <b>{idx}</b> من بيانات الاختبار '
            f'&nbsp;|&nbsp; القيمة الحقيقية: <b>{y_test.iloc[idx]}</b></div>'
        ))


def predict(_):
    with output:
        clear_output()
        raw = textbox.value.strip()
        if not raw:
            display(HTML('<div class="warn-box">⚠️ من فضلك ادخل القيم أو دوس "عينة عشوائية"</div>'))
            return
        try:
            values = [float(v.strip()) for v in raw.split(',') if v.strip() != '']
        except ValueError:
            display(HTML('<div class="warn-box">⚠️ تأكد إن كل القيم أرقام صحيحة مفصولة بفواصل</div>'))
            return

        if len(values) != N_FEATURES:
            display(HTML(
                f'<div class="warn-box">⚠️ لازم تدخل بالظبط {N_FEATURES} قيمة، إنت دخلت {len(values)}</div>'
            ))
            return

        arr_df = pd.DataFrame([values], columns=feature_names)
        pred = model.predict(arr_df)[0]
        proba = model.predict_proba(arr_df)[0]
        classes = list(model.classes_)
        p_r = proba[classes.index('R')]
        p_m = proba[classes.index('M')]

        is_rock = pred == 'R'
        box_class = 'result-rock' if is_rock else 'result-mine'
        icon = '🪨' if is_rock else '💣'
        label = 'صخرة (Rock)' if is_rock else 'لغم (Mine)'

        html = f"""
        <div class="result-box {box_class}">
            <div class="result-label">{icon} &nbsp; النتيجة: {label}</div>
            <div class="prob-row">
                <div class="prob-name">صخرة (R)</div>
                <div class="prob-bar-bg"><div class="prob-bar-fill-r" style="width:{p_r*100:.1f}%"></div></div>
                <div class="prob-val">{p_r*100:.1f}%</div>
            </div>
            <div class="prob-row">
                <div class="prob-name">لغم (M)</div>
                <div class="prob-bar-bg"><div class="prob-bar-fill-m" style="width:{p_m*100:.1f}%"></div></div>
                <div class="prob-val">{p_m*100:.1f}%</div>
            </div>
        </div>
        """

        display(HTML(html))


def clear_all(_):
    textbox.value = ''
    with output:
        clear_output()


random_btn.on_click(load_random_sample)
predict_btn.on_click(predict)
clear_btn.on_click(clear_all)
```
