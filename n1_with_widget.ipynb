## 🎛️ واجهة تفاعلية لتشغيل الموديل (Interactive Widget)
شغّل الخلية دي بعد ما تشغّل كل خلايا التدريب اللي فوق (لازم تكون متغيرات `model`, `x`, `x_test`, `y_test` موجودة في الذاكرة).

```python
# تأكد إن مكتبة ipywidgets متاحة
# لو مش متاحة شغل: pip install ipywidgets

import ipywidgets as widgets
from IPython.display import display, clear_output
import numpy as np
import pandas as pd

feature_names = list(x.columns)  # اسماء الاعمدة الـ 60 اللي اتدرب عليها الموديل

title = widgets.HTML("<h2>🔍 Sonar Signal Classifier — Rock (R) vs Mine (M)</h2>")

instructions = widgets.HTML(
    "<p>الصق 60 قيمة (مفصولة بفاصلة) تمثل قراءات السونار، أو دوس على "
    "\"عينة عشوائية\" عشان تجرب بيانات حقيقية من الـ test set.</p>"
)

textbox = widgets.Textarea(
    value='',
    placeholder='مثال: 0.02, 0.0371, 0.0428, ... (60 قيمة)',
    layout=widgets.Layout(width='750px', height='100px')
)

random_btn = widgets.Button(description="🎲 عينة عشوائية", button_style='info')
predict_btn = widgets.Button(description="🚀 تنبؤ", button_style='success')
clear_btn = widgets.Button(description="🗑️ مسح", button_style='warning')
output = widgets.Output()


def load_random_sample(_):
    idx = np.random.randint(0, len(x_test))
    sample = x_test.iloc[idx]
    textbox.value = ', '.join(map(str, sample.values))
    with output:
        clear_output()
        print(f"✅ تم تحميل عينة رقم {idx} من بيانات الاختبار (القيمة الحقيقية: {y_test.iloc[idx]})")


def predict(_):
    with output:
        clear_output()
        raw = textbox.value.strip()
        if not raw:
            print("⚠️ من فضلك ادخل قيم أو دوس عينة عشوائية")
            return
        try:
            values = [float(v.strip()) for v in raw.split(',') if v.strip() != '']
        except ValueError:
            print("⚠️ تأكد إن كل القيم أرقام صحيحة مفصولة بفواصل")
            return

        if len(values) != len(feature_names):
            print(f"⚠️ لازم تدخل بالظبط {len(feature_names)} قيمة، إنت دخلت {len(values)}")
            return

        arr_df = pd.DataFrame([values], columns=feature_names)
        pred = model.predict(arr_df)[0]
        proba = model.predict_proba(arr_df)[0]
        classes = list(model.classes_)

        label = "🪨 صخرة (Rock)" if pred == 'R' else "💣 لغم (Mine)"
        print(f"النتيجة: {label}\n")
        for cls, p in zip(classes, proba):
            name = "صخرة (R)" if cls == 'R' else "لغم (M)"
            print(f"احتمالية {name}: {p:.2%}")


def clear_all(_):
    textbox.value = ''
    with output:
        clear_output()


random_btn.on_click(load_random_sample)
predict_btn.on_click(predict)
clear_btn.on_click(clear_all)

display(
    title,
    instructions,
    textbox,
    widgets.HBox([random_btn, predict_btn, clear_btn]),
    output,
)
```
